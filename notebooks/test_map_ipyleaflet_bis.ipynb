```python
from ipyleaflet import Map, GeomanDrawControl

m = Map(center=(30, 360), zoom=5)

draw_control = GeomanDrawControl()
draw_control.polyline =  {}
draw_control.circlemarker = {}
draw_control.circle = {}
draw_control.polygon = {
    "pathOptions": {
        "fillColor": "#fca45d",
        "color": "#fca45d",
        "fillOpacity": 0.30
    }
}
draw_control.rectangle = {
    "pathOptions": {
        "fillColor": "#fca45d",
        "color": "#fca45d",
        "fillOpacity": 0.30
    }
}

m.add(draw_control)

m
```

```python
import geojson

AOI_file = "ipyleaflet_AOI_bis.geojson"
data = draw_control.data
print(data)

data_file = open(AOI_file,'w')
geojson.dump(data, data_file)
data_file.close()
```

```python
import geojson
import geopandas as gpd
from shapely.geometry import box, Point, Polygon, MultiPolygon
from shapely import from_geojson, normalize, transform

AOI_file = "ipyleaflet_AOI_bis.geojson"
data_file = open(AOI_file,'r')
data = geojson.load(data_file)
data_file.close()

list_geom = []
for i in range(len(data)):
    list_geom.append(transform(Polygon(data[i]['geometry']['coordinates'][0]), lambda x: x - [360, 0]))

aoi_geom = MultiPolygon(list_geom)

aoi_poly_geom = gpd.GeoDataFrame(index=[0], crs='epsg:4326', geometry=[aoi_geom]) 
edge = [
    [aoi_poly_geom.total_bounds[1],aoi_poly_geom.total_bounds[0]],
    [aoi_poly_geom.total_bounds[3],aoi_poly_geom.total_bounds[2]]
]
```

```python
import folium

m = folium.Map(location=[30,0], zoom_start=5)
folium.GeoJson(aoi_geom,color='red').add_to(m)
m.fit_bounds(edge)

m
```
