In the file __variables.py__ is present a list of possible map background that can be use in SOURCE.<br>
You can add new elements in the list or delete the element present !<br>
In this code you can see every specific map.<br>
<br>

```
map_flavours = [
  "OpenStreetMap",
  "Cartodb Positron",
  "Stadia.StamenToner",
  "Stadia.StamenTonerBackground",
  "Stadia.StamenTonerLite",
  "Stadia.StamenTerrainBackground",
  "Esri.WorldImagery",
  "Esri.WorldTerrain",
  "Esri.WorldShadedRelief",
  "Esri.WorldPhysical",
  "Esri.OceanBasemap",
  "Esri.WorldGrayCanvas",
  "CartoDB.Positron",
  "CartoDB.PositronNoLabels",
  "CartoDB.DarkMatter",
  "CartoDB.DarkMatterNoLabels",
  "CartoDB.Voyager",
  "CartoDB.VoyagerNoLabels",
  "GeoportailFrance.orthos",
  "USGS.USTopo"
]
```
<br>
For the use in SOURCE you mast use the index of the element in the list, for example 0 = OpenStreetMap, 1 = Cartodb Positron and so on for the other element of the list.<br>
